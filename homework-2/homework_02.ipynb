```python
def validate_email(email_string):
    email = email_string.strip().lower()
    if (email.count('@') != 1): return False
    if (email.startswith("@") or email.endswith("@")): return False
    
    local_part, domain = email.split('@')

    if ('.' not in domain): return False
    if (domain.startswith(".") or domain.endswith(".")): return False
    
    return True

print(validate_email("test@mail.ru"))
print(validate_email("@testmail.ru"))
print(validate_email("test@@mail.ru"))
print(validate_email("test@mailru"))
```

**Задача 5: Детектор кодировок**

Напишите функцию `check_encoding_decoding(text)`, которая принимает строку с кириллицей. Функция должна:
1.  Закодировать эту строку в `utf-8` и в `windows-1251`.
2.  Попытаться декодировать байты `windows-1251` обратно в строку, используя `utf-8`.
3.  Вернуть кортеж из трех элементов: исходный текст, корректно декодированный текст и некорректно декодированный текст.


```python
def encoding_decoding(text):
    
    s01 = text.encode('utf-8')
    s02 = text.encode('windows-1251')

    s12 = s02.decode('windows-1251')
    try:
        s22 = s02.decode('utf-8')
    except UnicodeDecodeError:
        s22 = s02.decode('utf-8', errors='replace')
    
    return (text, s12, s22)

encoding_decoding("привет")
```

**Задача 6: Парсер команд робота**

Напишите функцию `parse_robot_command(command)`, которая принимает одну строку. Используя `match/case`, разберите команду:
*   Команды `"вперед"` или `"назад"` должны возвращать `"Робот движется..."`.
*   Команды `"налево"` или `"направо"` должны возвращать `"Робот поворачивает..."`.
*   Команда `"стоп"` должна возвращать `"Робот остановлен."`.
*   Любая другая команда должна возвращать `"Неизвестная команда."`.

Используйте оператор `|` для группировки похожих команд.


```python
def parse_robot_command(command):
    match command:
        case ("вперед" | "назад"):
            s = "Робот движется..."
        case ("налево" | "направо"):
            s = "Робот поворачивает..."
        case "стоп":
            s = "Робот остановлен."
        case _:
            s = "Неизвестная команда."
            
    return s

print(parse_robot_command("вперед"))
print(parse_robot_command("направо"))
print(parse_robot_command("стоп"))
print(parse_robot_command("иди"))
```

Напишите функцию `is_valid_pin(pin_code)`, которая проверяет, что строка с PIN-кодом состоит **только** из цифр. Длина PIN-кода должна быть ровно 4 или 6 символов. Функция должна вернуть `True` или `False`.
**Обязательное условие:** Для проверки на наличие не-цифровых символов используйте цикл `for` с блоком `else`.


```python
def is_valid_pin(pin_code):
    n=0
    s = False
    for i in pin_code:
        if (i<'0' or i>'9'): break
        n+=1
    else:
        s = True
    if(s and (n==4 or n==6)): return True
    else: return False

print(is_valid_pin("123"))
print(is_valid_pin("123a"))
print(is_valid_pin("1233r"))
print(is_valid_pin("123222"))
    

```

**Задача 8: Фильтр "стоп-символов"**

Напишите функцию `filter_text(text, stop_chars)`, которая принимает основной текст и строку "стоп-символов". Функция должна вернуть новую строку, в которой отсутствуют все символы, перечисленные в `stop_chars`.
**Обязательное условие:** Используйте цикл `for` с оператором `continue`.


```python
def filter_text(text,stop_chars):
    s = text
    for i in stop_chars:
        if(s.count(i)==0): continue
        else: s = s.replace(i,"")
    return s

print(filter_text("Hello world!","hol"))
print(filter_text("Hello world!",""))
```

**Задача 9: Шифр Цезаря с `enumerate`**

Напишите функцию `caesar_cipher_encoder(text)`, которая "шифрует" текст по следующему правилу: символы на четных индексах сдвигаются на 1 вперед по алфавиту (`a`->`b`, `b`->`c`), а символы на нечетных индексах сдвигаются на 1 назад (`c`->`b`, `b`->`a`). Пробелы и знаки препинания не меняются.
**Обязательное условие:** Используйте `enumerate` для получения индекса и символа в цикле.
*(Подсказка: функции `ord()` и `chr()` могут быть полезны для работы с кодами символов).*


```python
def caesar_cipher_encoder(text):
    s = []
    for index, c in enumerate(text):
        if(c==' ' or c == '.' or c == ',' or c == '!' or c == '?' or c == ':' or c == ';' or c == '"'): 
            s.append(c)
            continue
        if(index % 2):
            s.append(chr(ord(c)+1))
        else:
            s.append(chr(ord(c)-1))
            
    return "".join(s)

print(caesar_cipher_encoder("ccc. ff"))
```

**Задача 10: Комплексный валидатор пароля**

Напишите функцию `is_strong_password(password)`, которая проверяет пароль на надежность и возвращает `True` или `False`.
Надежный пароль должен соответствовать **всем** следующим критериям:
1.  Длина не менее 10 символов.
2.  Содержит хотя бы одну букву в верхнем регистре (`A-Z`).
3.  Содержит хотя бы одну букву в нижнем регистре (`a-z`).
4.  Содержит хотя бы одну цифру (`0-9`).
5.  Содержит хотя бы один из спецсимволов: `!@#$%^&*()`.


```python
def is_strong_password(password):
    if (len(password)<10): return False

    s = [False,False,False,False]
    special_chars = "!@#$%^&*()"

    for index, c in enumerate(password):
        if (c.isupper()):s[0]=True
        elif (c.islower()):s[1]=True
        elif (c.isdigit()):s[2]=True
        elif (c in special_chars): s[3]=True

        if (s[0] and s[1] and s[2] and s[3]): break

    return s[0] and s[1] and s[2] and s[3]

print(is_strong_password("qW12#psdofksfsfdk"))
print(is_strong_password("qW12psdofksfsfdk"))
print(is_strong_password("qW#psdofksfsfdk"))
print(is_strong_password("qW12#"))
print(is_strong_password("q12#psdofksfsfdk"))
```
