## Задание 1: Счетчик слов

Напишите функцию, которая читает текстовый файл и возвращает словарь с количеством:
- строк
- слов
- символов (без пробелов)

**Файл:** `data/sample_text.txt`

**Требования:**
- Обработайте исключение `FileNotFoundError`
- Используйте `encoding='utf-8'`
- Верните словарь с ключами: `'lines'`, `'words'`, `'chars'`

```python
def f(path):
    d=dict()
    try:
        with open(path,'r', encoding='utf-8') as f:
            content = f.read()
        
        lines = content.splitlines()
        d['lines'] = len(lines)
        d['words'] = len(content.split())
        d['chars'] = len(content.replace(" ",""))

    except FileNotFoundError:
        print('File not found')
    else:
        print("File read successfully")
    return d

print(f('data/sample_text.txt'))
print()
print(f('data/sample.txt'))
```

## Задание 3: Поиск в файле

Напишите функцию, которая ищет в текстовом файле все строки, содержащие заданное слово (регистронезависимо), и возвращает список этих строк.

**Файл:** `data/sample_text.txt`

**Требования:**
- Поиск должен быть регистронезависимым
- Верните список найденных строк
- Обработайте `FileNotFoundError`

```python
def f(path,word):
    res = []
    try:
        with open(path,'r', encoding='utf-8') as f:
            content = f.readlines()
        for line in content:
            if word.lower() in line.lower():
                res.append(line)

    except FileNotFoundError:
        print('File not found')
    else:
        print("File read successfully")
    return res

print(f('data/sample_text.txt','python'))
print(f('data/sample_text.txt','Barcelona'))
print()
print(f('data/sample.txt','Barcelona'))
```

## Задание 5: Объединение файлов

Напишите функцию, которая объединяет несколько текстовых файлов в один. Перед содержимым каждого файла добавьте заголовок вида `=== имя_файла.txt ===`.

**Файлы:** `data/file1.txt`, `data/file2.txt`, `data/file3.txt`

- Верните количество успешно объединенных файлов
- Добавляйте заголовок перед содержимым каждого файла

```python
def f(files,new_file):
    s=0
    for name in files:
        try:
            with open(name,'r') as f1:
                content = f1.read()
        except FileNotFoundError:
            print(name,'not found')
        else:
            with open(new_file,'a') as f2:
                f2.write(name)
                f2.write('\n')
                f2.write(content)
                f2.write('\n')
            print(name,"read successfully")
            s+=1
    return s

print(f(['data/file1.txt', 'data/file2.txt', 'data/file3.txt'],'test5.txt'))
print(f(['data/file19.txt', 'data/file2.txt', 'data/file3.txt'],'test5-2.txt'))
```

## Задание 6: Фильтрация CSV

Напишите функцию, которая фильтрует строки CSV файла по заданному условию и сохраняет результат в новый CSV файл.

**Файл:** `data/employees.csv`

**Требования:**
- Используйте `csv.DictReader` и `csv.DictWriter`
- Третий параметр - функция-фильтр (lambda)
- Верните количество строк, прошедших фильтр
- **ОБЯЗАТЕЛЬНО:** Напишите тесты для этой функции


```python
import csv
import pytest

def f(name_input,name_output,filter):
    n = 0
    try:
        s1=[]
        with open(name_input,'r',encoding='utf-8') as f1:
            reader = csv.DictReader(f1)
            for row in reader:
                s1.append(row)

    except FileNotFoundError:
        print(name_input,'not found')
    else:
        s2=[row for row in s1 if filter(row)]
        n = len(s2)

        with open(name_output,'w',newline='', encoding='utf-8') as f2:
            writer = csv.DictWriter(f2,fieldnames=reader.fieldnames)
            writer.writeheader()
            for row in s2:
                writer.writerow(row)

    return n

print(f('data/employees.csv','data/employees_with_filter.csv',lambda row: int(row['age'])>30))
print(f('data/employees.csv','data/employees_with_filter.csv',lambda row: row['city'] == 'Москва'))

# запускать тесты через терминал из нового файла с расширением .py
# pytest .\test_task6.py -v

@pytest.mark.parametrize("name_input,name_output,filter,expected",[
    ('data/employees.csv','data/employees_with_filter.csv',lambda row: int(row['age']) > 30,3),
    ('data/employees.csv','data/employees_with_filter2.csv',lambda row: int(row['age']) > 40,1),
    ('data/employees.csv','data/employees_with_filter2.csv',lambda row: row['city'] == 'Москва',3),
])

def test_f(name_input,name_output,filter,expected):
    assert f(name_input,name_output,filter) == expected
        
```

Напишите функцию, которая читает JSON файл с товарами, применяет скидку к товарам дороже заданной цены, и сохраняет результат в новый JSON файл.

**Файл:** `data/products.json`

**Требования:**
- Функция принимает: путь к файлу, минимальную цену, процент скидки, путь для сохранения
- Примените скидку только к товарам с `price >= min_price`
- Сохраните результат с `indent=2` и `ensure_ascii=False`
- Верните количество товаров, к которым была применена скидка
- Обработайте `FileNotFoundError` и `json.JSONDecodeError`

```python
import json

def f(name_input,name_output,min_price,discount):
    try:
        with open(name_input,'r',encoding='utf-8') as f1:
            data = json.load(f1)
    except FileNotFoundError:
        print(name_input,'not found')
    except json.JSONDecodeError:
        print('json.JSONDecodeError')
    else:
        for row in data:
            if (row['price']>=min_price):
                row['price'] *= (1-discount/100.0)

        with open(name_output,'w',encoding='utf-8') as f2:
            json.dump(data,f2,indent=2,ensure_ascii=False)

f('data/incorrect_json.json','data/task7.json',1000,50)
f('data/products.json','data/task7.json',1000,50)
    
```

## Задание 8: Копирование файлов

Напишите функцию, которая копирует файлы определенного расширения из одной директории в другую.

**Требования:**
- Используйте `pathlib.Path` и `.glob()`
- Используйте `shutil.copy()`
- Создавайте целевую директорию, если её нет
- Верните список имен скопированных файлов

```python
from pathlib import Path
import shutil

def f(dir_from,dir_to,format):
    path_from = Path(dir_from)
    path_to = Path(dir_to)
    path_to.mkdir(parents=True, exist_ok=True)
    
    copied=[]
    for file in path_from.glob(format):
        if file.is_file():
            shutil.copy(file,path_to)
            copied.append(file.name)
    return copied

f('data','data2','*.txt')
```

## Задание 9: Анализатор логов

Напишите две функции для анализа файла логов.

**Формат:** `[2025-10-21 10:15:30] INFO: Application started`

**Функция 1:** Подсчитывает количество записей каждого уровня (INFO, WARNING, ERROR)
**Функция 2:** Находит все записи с уровнем ERROR

**Файл:** `data/application.log`

**Требования:**
- Используйте регулярные выражения (модуль `re`)
- **ОБЯЗАТЕЛЬНО:** Напишите тесты для каждой функции

```python
import re
import pytest

def g1(name_input):
    s={
        "INFO": 0,
        "WARNING": 0,
        "ERROR": 0
    }

    with open(name_input,'r',encoding='utf-8') as f:
        for line in f:
            match = re.search(r'\]\s+(\w+):', line)
            if match:
                level = match.group(1)
                s[level]+=1
    
    return s

def g2(name_input):
    s=[]

    with open(name_input,'r',encoding='utf-8') as f:
        for line in f:
            match = re.search(r'ERROR:\s*(.*)', line)
            if match:
                s.append(match.group(1))
    
    return s

print(g1('data/application.log'))
print(g1('data/file1.txt'))
print(g2('data/application.log'))
print(g2('data/file1.txt'))

# запускать тесты через терминал из нового файла с расширением .py
# pytest .\test_task9.py -v

@pytest.mark.parametrize("name_input, expected",[
    ('data/application.log',{'INFO': 12, 'WARNING': 4, 'ERROR': 4}),
    ('data/file1.txt',{'INFO': 0, 'WARNING': 0, 'ERROR': 0}),
])

def test_g1(name_input,expected):
    assert g1(name_input) == expected


@pytest.mark.parametrize("name_input, expected",[
    ('data/application.log',['Connection to database failed', 'File not found: config.json', 'Invalid user input', 'Failed to send notification']),
    ('data/file1.txt',[]),
])

def test_g2(name_input,expected):
    assert g2(name_input) == expected
```

## Задание 10: Конвертер CSV → JSON

Напишите две функции:
1. Конвертирует строку в нужный тип: `''` → `None`, `'true'/'false'` → `bool`, числа → `int/float`
2. Читает CSV и конвертирует в JSON с преобразованием типов

**Файл:** `data/mixed_data.csv`

**Требования:**
- Используйте модули `csv` и `json`
- Сохраните JSON с `indent=2` и `ensure_ascii=False`
- **ОБЯЗАТЕЛЬНО:** Напишите тесты для функций

```python
import csv
import json
import pytest

def g1(value):
    if value == '':
        return None
    elif value.lower() == 'true':
        return True
    elif value.lower() == 'false':
        return False
    
    try:
        return int(value)
    except ValueError:
        pass
    
    try:
        return float(value)
    except ValueError:
        pass
    
    return value



def g2(input,output):
    data=[]
    with open(input,'r',encoding='utf-8') as f1:
        reader = csv.DictReader(f1)
    
        for row in reader:
            s={}
            for key,value in row.items():
                s[key] = g1(value)
            data.append(s)

    with open(output,'w',encoding='utf-8') as f2:
        json.dump(data,f2,indent=2,ensure_ascii=False)
    return data

print(g2('data/mixed_data.csv','data/mixed_data2.json'))

print(g1('321'))
print(g1('3.1'))
print(g1('true'))
print(g1('false'))
print(g1('Книга'))


@pytest.mark.parametrize("value, expected",[
    ('321',321),
    ('3.1',3.1),
    ('true',True),
    ('false',False),
    ('Книга','Книга'),
])

def test_g1(value,expected):
    assert g1(value) == expected


@pytest.mark.parametrize("input,output, expected",[
    ('data/mixed_data.csv','data/mixed_data2.json',[{'product': 'Ноутбук', 'price': 50000, 'in_stock': True, 'rating': 4.5}, {'product': 'Мышь', 'price': 500, 'in_stock': True, 'rating': 4.2}, {'product': 'Книга', 'price': 1200, 'in_stock': False, 'rating': 4.8}, {'product': 'Клавиатура', 'price': 1500, 'in_stock': True, 'rating': 4.0}]),
])

def test_g2(input,output,expected):
    assert g2(input,output) == expected
```
