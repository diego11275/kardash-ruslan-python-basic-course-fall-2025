### Задача 2: Класс `User` с валидацией пароля
Создайте класс `User` для представления пользователя.
*   **Атрибуты:** `username` (публичный), `__password` (приватный).
*   **Методы:**
    *   `__init__`: принимает `username` и `password`.
    *   `set_password(new_password)`: устанавливает новый пароль. Пароль должен быть не короче 8 символов. Если пароль короче, вызовите исключение `ValueError`.
    *   `check_password(password_to_check)`: возвращает `True`, если переданный пароль совпадает с паролем пользователя, и `False` в противном случае.
    *   *Примечание:* Прямого способа "получить" пароль (`get_password`) быть не должно, чтобы продемонстрировать инкапсуляцию.

```python
class User:
    def __init__(self,username,password):
        self.username = username
        self.__password = password
    
    def set_password(self,new_password):
        if (len(new_password)<8):
            raise ValueError("Пароль должен быть не короче 8 символов.")
        self.__password=new_password

    def check_password(self,password_to_check):
        res = False
        if(self.__password==password_to_check):
            res = True
        return res
    
s1 = User("Рыбак","12345")
print(s1.username)
print(s1.check_password("12345"))
s1.set_password("12345678")
print(s1.check_password("12345"))
print(s1.set_password("123"))
```

### Задача 3: Класс `StringUtils` со статическими методами
Создайте класс-утилиту `StringUtils`, который будет содержать статические методы для работы со строками. Этот класс не должен иметь `__init__` или атрибутов экземпляра.
*   **Методы:**
    *   `@staticmethod reverse_string(s)`: принимает строку `s` и возвращает её в перевернутом виде.
    *   `@staticmethod is_palindrome(s)`: принимает строку `s` и возвращает `True`, если она является палиндромом (читается одинаково в обоих направлениях), и `False` в противном случае.
    *   `@staticmethod count_vowels(s)`: принимает строку `s` и возвращает количество гласных букв в ней (a, e, i, o, u, как строчных, так и прописных).

```python
class StringUtils:
    @staticmethod
    def reverse_string(s):
        return s[::-1]
    
    @staticmethod
    def is_palindrome(s):
        res = False
        if (s==s[::-1]):
            res = True
        return res
    
    @staticmethod
    def count_vowels(s):
        count = s.count('a') + s.count('A')
        count += s.count('e') + s.count('E')
        count += s.count('i') + s.count('I')
        count += s.count('o') + s.count('O')
        count += s.count('u') + s.count('U')
        return count

print(StringUtils.reverse_string("12345"))
print(StringUtils.is_palindrome("12345"))
print(StringUtils.is_palindrome("121"))
print(StringUtils.count_vowels("Aa Ee Ii Oo Uu qwrtypsdfghjklzxcvbnm"))
```

### Задача 4: Класс `Person` со свойством `age`
Создайте класс `Person`, который хранит возраст человека.
*   **Атрибуты:** `_name` (защищенный), `_age` (защищенный).
*   **Свойства (properties):**
    *   Создайте свойство `age`.
    *   Геттер (`@property`) для `age` должен возвращать значение `_age`.
    *   Сеттер (`@age.setter`) должен проверять, что устанавливаемый возраст находится в диапазоне от 0 до 120. Если возраст некорректен, вызовите исключение `ValueError`.
    *   Создайте свойство `is_adult` (только для чтения), которое возвращает `True`, если возраст 18 лет и больше, и `False` в противном случае.

```python
class Person:
    def __init__(self,name,age):
        self._name = name
        self._age = age
    
    @property
    def age(self):
        return self._age
    
    @age.setter
    def age(self,new_age):
        if not (0 <= new_age <= 120):
            raise ValueError("Устанавливаемый возраст должен находится в диапазоне от 0 до 120")
        self._age = new_age
    
    @property
    def is_adult(self):
        res = True
        if(self._age<18):
            res = False
        return res
    
s1 = Person("Mike",23)
print(s1.age)
s1.age = 10
print(s1.age)
print(s1.is_adult)
s1.age = -10
```

### Задача 6: Система уведомлений (абстрактные классы)
Создайте систему для отправки уведомлений, используя абстрактные классы.
*   Создайте абстрактный класс `Notification` с помощью модуля `abc`. В нём должен быть один абстрактный метод `send(message)`.
*   Создайте три конкретных класса, наследующих от `Notification`:
    1.  `EmailNotification`: в конструкторе принимает `recipient_email`. Метод `send` выводит "Отправка Email на [адрес]: [сообщение]".
    2.  `SMSNotification`: в конструкторе принимает `phone_number`. Метод `send` выводит "Отправка SMS на [номер]: [сообщение]".
    3.  `PushNotification`: в конструкторе принимает `device_id`. Метод `send` выводит "Отправка Push-уведомления на устройство [ID]: [сообщение]".
*   Напишите функцию, которая принимает список объектов-уведомлений и сообщение, а затем вызывает метод `send` для каждого из них.

```python
from abc import ABC,abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(message):
        pass

class EmailNotification(Notification):
    def __init__(self,recipient_email):
        self.recipient_email = recipient_email
    
    def send(self,message):
        print(f"Отправка Email на [{self.recipient_email}]: [{message}]")

class SMSNotification(Notification):
    def __init__(self,phone_number):
        self.phone_number = phone_number
    
    def send(self,message):
        print(f"Отправка SMS на [{self.phone_number}]: [{message}]")

class PushNotification(Notification):
    def __init__(self,device_id):
        self.device_id = device_id
    
    def send(self,message):
        print(f"Отправка PUSH-уведомления на устройство [{self.device_id}]: [{message}]")

def send_notifications(notifications,message):
    for notification in notifications:
        notification.send(message)

s = [EmailNotification("mike@mail.ru"),SMSNotification(7352315),PushNotification(10)]
send_notifications(s,"Hello world!")
```

### Задача 8: Магические методы для класса `Vector`
Создайте класс `Vector` для представления 2D-вектора.
*   **Атрибуты:** `x`, `y`.
*   Реализуйте следующие магические методы:
    *   `__add__(self, other)`: для сложения векторов (`v1 + v2`).
    *   `__sub__(self, other)`: для вычитания векторов (`v1 - v2`).
    *   `__mul__(self, scalar)`: для умножения вектора на скаляр (число) (`v1 * 3`).
    *   `__str__` и `__repr__` для красивого вывода.

```python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def __add__(self,other):
        if not isinstance(other, Vector):
            raise TypeError("Можно складывать только векторы")
        return Vector(self.x+other.x,self.y+other.y)
    
    def __sub__(self,other):
        if not isinstance(other, Vector):
            raise TypeError("Можно вычитать только векторы")
        return Vector(self.x-other.x,self.y-other.y)

    def __mul__(self,scalar):
        if not isinstance(scalar, (int, float)):
            raise TypeError("Можно умножать только на число")
        return Vector(self.x*scalar,self.y*scalar)
    
    def __str__(self):
        return f"({self.x},{self.y})"
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    

v1 = Vector(2, 3)
v2 = Vector(1, 4)
    
print("v1 =", v1)
print("v2 =", v2)
    
v3 = v1 + v2
print("v1 + v2 =", v3)
v4 = v1 - v2
print("v1 - v2 =", v4)
v5 = v1 * 3
print("v1 * 3 =", v5)

# Вывод в списке (использует __repr__)
vectors = [v1, v2, v3]
print("Список векторов:", vectors)
```

*   **Функциональность `TaskManager`:**
    *   Добавление новой задачи.
    *   Просмотр списка всех задач.
    *   Изменение статуса задачи по её описанию или ID.

    *   Добавьте возможность сохранять задачи в файл (например, JSON) и загружать их оттуда.

```python
import json
class Task:
    def __init__(self,description,due_date,status="в ожидании"):
        self.description = description
        self.due_date = due_date
        self.status = status
    
    def __str__(self):
        return f"Задача: '{self.description}', срок: {self.due_date}, статус: {self.status}"
    
    def __repr__(self):
        return f"Task('{self.description}', '{self.due_date}', '{self.status}')"
    
    def __eq__(self,other):
        return self.description == other.description and self.due_date == other.due_date and self.status == other.status

    def __hash__(self):
        return hash((self.description, self.due_date, self.status))

    def to_dict(self):
        return {
            'description': self.description,
            'due_date': self.due_date,
            'status': self.status
        }

class TaskManager:
    def __init__(self,tasks_list):
        self.tasks = set(tasks_list)
    
    def add(self,task):
        if not isinstance(task, Task):
            raise TypeError("Можно добавлять только объект типа Task")
        self.tasks.add(task)

    def show_tasks(self):
        print(self.tasks)
    
    def __str__(self):
        return f"{self.tasks}"
    
    def __repr__(self):
        return f"Tasks:'\n'{self.tasks}"
    
    def change_status(self,identifier,new_status):
        if(type(identifier)==int):
            for task in self.tasks:
                if (id(task) == identifier):
                    task.status = new_status        
        else:
            for task in self.tasks:
                if (task.description == identifier):
                    task.status = new_status
    
    def remove(self,task):
        if not isinstance(task, Task):
            raise TypeError("Можно удалять только объект типа Task")
        self.tasks.discard(task)
    
    def write_in_file(self,output):
        data = []
        for task in self.tasks:
            data.append(task.to_dict())
        with open(output,'w',encoding='utf-8') as f:
            json.dump(data,f,indent=2,ensure_ascii=False)
    
    def load_from_file(self,input):
        with open(input,'r',encoding='utf-8') as f:
            data = json.load(f)
        for elem in data:
            self.tasks.add(Task(elem['description'],elem['due_date'],elem['status']))


tasks_list = [
    Task("Записаться к стоматологу", "2024-12-15"),
    Task("Сделать презентацию для работы", "2024-12-10", "в процессе"),
]
s = TaskManager(tasks_list)
print(s)
s.add(Task("Купить продукты на неделю", "2024-12-08", "в ожидании"))
s.show_tasks()
s.change_status("Записаться к стоматологу","завершено")
s.remove(Task("Купить продукты на неделю", "2024-12-08", "в ожидании"))
s.show_tasks()

tasks_list2 = [
    Task("Оплатить коммунальные услуги", "2024-12-05", "завершено"),
    Task("Подготовить отчет по проекту", "2024-12-20", "в процессе"),
    Task("Забрать посылку с почты", "2024-12-09", "в ожидании"),
    Task("Записаться на курсы английского", "2024-12-25", "в ожидании"),
    Task("Починить кран на кухне", "2024-12-12", "в процессе"),
    Task("Сходить в спортзал", "2024-12-06", "в ожидании")
]
    
TaskManager(tasks_list2).write_in_file("task9.1.json")
s.load_from_file("task9.1.json")
s.write_in_file("task9.2.json")
```

### Задача 10: Паттерн "Фабрика" для создания документов
Реализуйте паттерн "Фабричный метод" для создания различных типов документов.
*   Создайте абстрактный класс `Document` с абстрактным методом `display()`.
*   Создайте конкретные классы-документы: `PDFDocument`, `WordDocument`, `WebDocument`, каждый из которых реализует метод `display()` (например, выводит строку "Отображаю PDF-документ").
*   Создайте класс `DocumentFactory`, который имеет статический метод `create_document(doc_type)`. Этот метод на основе строки (`"pdf"`, `"word"`, `"web"`) возвращает экземпляр соответствующего класса документа.
*   Клиентский код должен использовать только фабрику для создания документов, не создавая их напрямую. Это позволит в будущем легко добавлять новые типы документов, не меняя код клиента.

```python
from abc import ABC,abstractmethod

class Document(ABC):
    @abstractmethod
    def display(self):
        pass

class PDFDocument(Document):
    def display(self):
        print("Отображаю PDF-документ")

class WordDocument(Document):
    def display(self):
        print("Отображаю Word-документ")

class WebDocument(Document):
    def display(self):
        print("Отображаю Web-документ")

class DocumentFactory:
    @staticmethod
    def create_document(doc_type):
        match doc_type:
            case "pdf":
                res = PDFDocument()
            case "word":
                res = WordDocument()
            case "web":
                res = WebDocument()
        return res

DocumentFactory.create_document("pdf").display()
DocumentFactory.create_document("word").display()
DocumentFactory.create_document("web").display()
```
