### Task 1: Temperature Analysis
Create an array of 365 random numbers representing daily temperatures (e.g., from -10 to 35).
Find:
1. The average temperature for the year.
2. The average temperature for the "summer" months (from day 150 to day 240).
3. The standard deviation of the temperature for the whole year.

````python
import numpy as np
arr = np.random.randint(-10,35,size=(365))
# print(arr)
print(f"Средняя температура за год: {np.mean(arr):.2f}°C")
print(f"Средняя температура за летние месяцы: {np.mean(arr[150:240]):.2f}°C")
print(f"Стандартное отклонение температуры за год: {np.std(arr):.2f}°C")
````

### Task 6: Sales Analysis
Given a 5x7 sales matrix (5 products, 7 days of the week), filled with random integers.
1. Calculate the total revenue of the store for the week.
2. Find the day of the week (column index) when the maximum volume of goods was sold.
3. Find the product (row index) that sold the worst (minimum total sales).

````python
sales = np.random.randint(0,2,size=(5,7))
prices = np.random.randint(10,20,size=(1,5))
print(sales,'\n\n',prices)

print(f"\n1. Общая выручка магазина за неделю: {np.sum(prices @ sales)}")

daily_totals = np.sum(sales, axis=0)
print(f"\n2. Продажи по дням недели: {daily_totals}")
max_day_index = np.argmax(daily_totals)
print(f"   Максимальный объем продаж был в день с индексом: {max_day_index}")

product_totals = np.sum(sales, axis=1)
print(f"\n3. Суммарные продажи по товарам: {product_totals}")
min_product_index = np.argmin(product_totals)
print(f"   Хуже всего продавался товар с индексом: {min_product_index}")
````

### Задача 7: Бонусные баллы
В группе 5 студентов, каждый сдал 3 экзамена. Результаты — матрица 5x3 (случайные числа от 50 до 100).

### Задача 8: Логи сервера
Дан массив строк, имитирующий лог-файл с кодами ответов сервера: `["200", "404", "200", "500", "200", "404"]`.
1. Преобразуйте этот массив в массив целых чисел.
2. Посчитайте процент успешных запросов (код 200) от общего числа запросов.

````python
lines = np.array(["200", "404", "200", "500", "200", "404"])
print(lines)

int_arr = lines.astype(np.int64)
print(int_arr)
print(f"Процент успешных запросов: {100*np.count_nonzero(int_arr == 200)/len(int_arr)}")
````

### Задача 9: Расстояния до точек
Есть 10 точек на плоскости (координаты x, y) — массив размером 10x2.
Найдите Евклидово расстояние от первой точки (индекс 0) до всех остальных точек.
Реализуйте это одной строкой, используя broadcasting и векторные операции.

````python
points = np.random.randint(-5,5, size=(10,2))
print(points)
print(np.sqrt(np.sum((points - points[0,0:2])**2,axis=1)))
````

### Задача 10: Поиск похожего пользователя (Рекомендательная система)
Дана матрица оценок 5 пользователей для 4 фильмов. Оценки — от 1 до 5, 0 означает, что фильм не просмотрен.
```python
ratings = np.array([
    [5, 4, 0, 1], # User 0
    [1, 0, 5, 4], # User 1
    [5, 4, 1, 0], # User 2
    [0, 5, 4, 5], # User 3
    [1, 1, 5, 4]  # User 4
])
```
Найдите пользователя (его индекс), чьи вкусы наиболее похожи на вкусы первого пользователя (User 0).
Для этого:
1. Вычислите разницу между оценками User 0 и всех остальных пользователей.
2. Найдите пользователя с минимальной суммарной разницей оценок (исключая самого User 0).

````python
ratings = np.array([
    [5, 4, 0, 1], # User 0
    [1, 0, 5, 4], # User 1
    [5, 4, 1, 0], # User 2
    [0, 5, 4, 5], # User 3
    [1, 1, 5, 4]  # User 4
])

# difference = - ratings + ratings[0,:]
difference = np.abs(ratings[0,:] - ratings)
print(difference)
print(np.sum(difference[1:,[0,1,3]],axis = 1))
print(np.argmin(np.sum(difference[1:,[0,1,3]],axis = 1)))

print(f"Наиболее похож: User {1+np.argmin(np.sum(difference[1:,[0,1,3]],axis = 1))}")
````
