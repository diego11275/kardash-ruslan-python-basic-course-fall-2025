### Задача 3
Найдите медианную стоимость билета (Fare) для каждого класса (Pclass). В каком классе самый большой разброс цен (максимальная - минимальная)?

### Задача 4
Для каждой комбинации класса (Pclass) и порта посадки (Embarked) посчитайте:
- Количество пассажиров
- Процент выживших
- Средний возраст

```python
groupby_data = df.groupby(['Pclass', 'Embarked']).aggregate({
    'Survived': ['sum','count'],
    'Age': 'mean'
}).round(1)
# print(groupby_data)

result_df = pd.DataFrame()
result_df['Passenger'] = groupby_data[('Survived','count')]
result_df['Survival_percent'] = (100*groupby_data[('Survived','sum')]/groupby_data[('Survived','count')]).round(1)
result_df['Mean_age'] = groupby_data[('Age','mean')]

print(result_df)
```

### Задача 5

### Задача 6
Создайте столбец `FamilySize = SibSp + Parch + 1` (размер семьи включая самого пассажира).

Найдите, при каком размере семьи выживаемость максимальная, а при каком — минимальная.

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

groupby_data = df.groupby(['FamilySize']).agg(
    Survived_sum=('Survived', 'sum'),
    Passenger_count=('Survived', 'count'))
groupby_data['Survival_percent'] = (100*groupby_data['Survived_sum']/groupby_data['Passenger_count']).round(1)

print(groupby_data,'\n')

print(f"Максимальная выживаемость при размере семьи: {groupby_data['Survival_percent'].idxmax()}")
print(f"Минимальная выживаемость при размере семьи: {groupby_data['Survival_percent'].idxmin()}")
```

### Задача 7
В столбце Age много пропусков. Заполните их **медианным возрастом** для соответствующей группы по полу (Sex) и классу (Pclass).

Например: пропуск у женщины из 1-го класса заполняется медианным возрастом женщин 1-го класса.

Сохраните результат в новый столбец `AgeFilled`.

```python
df['AgeMeanByGroup'] = df.groupby(['Sex', 'Pclass'])['Age'].transform('mean')
df['AgeFilled'] = df['Age'].fillna(df['AgeMeanByGroup'])
df = df.drop('AgeMeanByGroup', axis=1)

print(f"Пропуски до: {df['Age'].isnull().sum()}")
print(f"Пропуски после: {df['AgeFilled'].isnull().sum()}")

df.tail(3)
```

### Задача 8
Создайте столбец `AgeGroup` с категориями:
- `'Child'`: возраст < 12
- `'Teen'`: 12 ≤ возраст < 18
- `'Adult'`: 18 ≤ возраст < 60
- `'Senior'`: возраст ≥ 60

Постройте сводную таблицу (pivot table): **процент выживших** для каждой комбинации возрастной группы и пола.

```python
categorize = lambda age: next(c for a, c in [(12, 'Child'), (18, 'Teen'), (60, 'Adult'), (float('inf'), 'Senior')] if age < a)
df['AgeGroup'] = df['AgeFilled'].apply(categorize)


groupby_data = df.groupby(['Sex', 'AgeGroup']).aggregate({
    'Survived': ['sum','count'],
})
groupby_data['Survival_percent'] = (100*groupby_data[('Survived','sum')]/groupby_data[('Survived','count')]).round(1)
groupby_data
```

### Задача 10
Проверьте гипотезу «Женщины и дети первыми».

Создайте столбец `Priority`:
- `'High'`: женщины или дети (возраст < 16)
- `'Low'`: остальные (мужчины 16+)

Для каждого класса (Pclass) посчитайте:
- Процент выживших с высоким приоритетом
- Процент выживших с низким приоритетом
- Разницу между ними

В каком классе принцип «женщины и дети первыми» соблюдался лучше всего?

```python
df['Priority'] = df.apply(lambda row: 'Low' if (row['Sex']=='male') and (row['AgeFilled'] >=16) else 'High',axis=1)

groupby_data = df.groupby(['Pclass', 'Priority']).agg(
    Survived_sum=('Survived', 'sum'),
    Passenger_count=('Survived', 'count'))
groupby_data['Survival_percent'] = (100*groupby_data['Survived_sum']/groupby_data['Passenger_count']).round(1)

print(f"\nПринцип «женщины и дети первыми» соблюдался лучше всего в классе: {groupby_data.xs('High', level='Priority')['Survival_percent'].idxmax()}\n")

groupby_data['Differents'] = 0.0
for id in groupby_data.index:
    if (id[1]=='High'):
        groupby_data.loc[id,'Differents'] = groupby_data.loc[id,'Survival_percent'] - groupby_data.loc[(id[0],'Low'),'Survival_percent']

print(groupby_data)
```
